```python
#Write a program that takes a number as input and prints whether it's positive, negative, or zero.
def check_number():
    number = int(input())
    if number > 0:
        print("positive")
        return
    elif number == 0:
        print("zero")
        return
    print("negative")
check_number()
```

```python
#Write a program that determines if a year entered by the user is a leap year. (Hint: A leap year is divisible by 4, except for century years which must be divisible by 400).
def check_year():
    year = int(input())
    if year % 100 == 0:
        if year % 400 == 0:
            print("leap year")
        else:
            print("no leap year")
    elif year % 4 == 0:
        print("leap year")
    else:
        print("no leap year")
check_year()
```

```python
#Create a simple calculator that takes two numbers and an operator (+, -, *, /) as input and performs the corresponding operation.
def check_number():
    number1 = int(input('Enter the first number'))
    operator= input('Enter the operator: +, -, *, / ')
    number2 = int(input('Enter the second number'))
    
    if  operator == '+':
        print(number1+number2)
        return
    elif operator == '-':
        print(number1-number2)
        return
    elif operator == '*':
        print(number1*number2)
        return
    elif operator == '/':
        print(number1/number2)
        return
check_number()
```

```python
#Write a program that categorizes a person by age: child (0-12), teenager (13-19), adult (20-64), or senior (65+).
def check_age():
    age = int(input())
    if age < 0:
        print("Invalid age")
    elif age <= 12:
        print("A child")
    elif age <= 19:
        print("A teenager")
    elif age <= 64:
        print("An adult")
    else:
        print("A senior")
check_age()
```

```python
#Create a program that determines if a triangle is equilateral (all sides equal), isosceles (two sides equal), or scalene (no sides equal) based on three side lengths.
def check_triangle():
    a = int(input("Enter side 1: "))
    b = int(input("Enter side 2: "))
    c = int(input("Enter side 3: "))
    
 
    if (a + b <= c) or (a + c <= b) or (b + c <= a):
        print("Not a valid triangle")
        return


    if a == b == c:
        print("Equilateral triangle")
    elif a == b or b == c or a == c:
        print("Isosceles triangle")
    else:
        print("Scalene triangle")

check_triangle()
```

```python
#Write a program that checks if a password meets basic criteria: at least 8 characters long and not equal to "password" or "123456".
def check_password():
    password = input("Enter password: ")
    
    if len(password) < 8:
        print("Password Must be at least 8 characters long.")
    elif password == "password" or password == "123456":
        print("Password rejected")
    else:
        print("Password accepted!")

check_password()
```

```python
#Create a program that categorizes a word as "short" (1-4 letters), "medium" (5-8 letters), or "long" (9+ letters).
def check_word():
    word = input("Enter a word: ")
    
    length = len(word)
    
    if length == 0:
        print("You didn't enter a word.")
    elif length <= 4:
        print("Short")
    elif length <= 8:
        print("Medium")
    else:
        print("Long")

check_word()
```

```python
#Write a program that checks if a string contains only digits, only letters, or a mix of both.
def check_string():
    text = input("Enter a string: ")
    
    if  len(text) == 0:
        print("The string is empty.")
    elif text.isdigit():
        print("Contains only digits")
    elif text.isalpha():
        print("Contains only letters")
    else:
        print("Contains a mix")

check_string()
```

```python
#Write a program that checks if a name starts with a vowel or consonant.
def check_string():
    name = input("Enter a name: ").strip()
    
    if not name:
        print("No name was entered.")
        return
        

    first_letter = name[0].lower()
    
    if not first_letter.isalpha():
        print("The name does not start with a valid letter.")

    elif first_letter in "aeiou":
        print("Starts with a vowel")
    else:
        print("Starts with a consonant")
    
    
    

check_string()
```

```python
#Write a program that checks if a sentence starts with a capital letter and ends with proper punctuation (., !, or ?).
def check_string():
    text = input("Enter a sentence: ")
    
    if not text:
        print("Empty text")
        return

    if text[0].isupper() and text[-1] in [".", "!", "?"]:
        print("Correctly formatted")
    else:
        print("Incorrect formatting")

check_string()
```
