```python
#Write a program that converts a user-provided number of days into years, weeks, and remaining days (assume 365 days per year).
NUMBER_OF_DAYS_IN_A_YEAR = 365
def calculate_days():
    print("Enter the number of days")
    days=int(input())
    years = days // NUMBER_OF_DAYS_IN_A_YEAR
    weeks = days % NUMBER_OF_DAYS_IN_A_YEAR // 7
    remaining_days = days % NUMBER_OF_DAYS_IN_A_YEAR % 7
    print(f"The number of years is {years},The number of weeks is {weeks},The number of remaining days is {remaining_days}")
    
calculate_days()
```

```python
#Write a program that asks the user for a radius and calculates both the area and circumference of a circle (use 3.14159 for π).
def calculate_area_circumfrence():
    PI=3.14159
    print("Enter the radius")
    radius=int(input())
    area = PI * (radius ** 2)
    circumference = 2 * PI * radius
    print(f"The area is {area}, and the circumference is {circumference}")
calculate_area_circumfrence()
```

```python
#Write a program that asks for two numbers and displays their sum, difference, product, quotient, remainder, and power (first number raised to the power of second number).
def calculate_variations():
    print("Enter the first number")
    num1=int(input())
    print("Enter the second number")
    num2=int(input())
    sum_numbers = num1 + num2
    difference = num1-num2
    product = num1*num2
    quotient = num1/num2
    remainder=num1%num2
    power = num1**num2
    print(f"The sum is {sum_numbers}, the difference is {difference}, the product is {product}, the quotient {quotient}, the remainder {remainder}, and power is {power}")
calculate_variations()
```

```python
#Write a program that takes a user-entered temperature in Celsius and converts it to Fahrenheit, displaying the result.
def convert_to_fahrenheit():
    print("Enter the temperature")
    temperature=int(input())
    Fahrenheit =  (temperature * 1.8)+32
    print(f"The temperature in fahrenheit  is {Fahrenheit}")
convert_to_fahrenheit()
```

```python
#Create a program that calculates the Body Mass Index (BMI) by asking for a person's weight (in kg) and height (in meters).
def convert_to_BMI():
    print("Enter your weight")
    weight=int(input())
    print("Enter your height")
    height=int(input())
    height = height**2
    BMI = weight /height
    print(f"The BMI is {BMI}")
convert_to_BMI()
```

```python
#Create a program that calculates the simple interest for a principal amount, interest rate, and time period (in years) entered by the user.
def calulate_simple_interest():
    print("Enter the principal amount:")
    principal = int(input())
    
    print("Enter the interest rate:")
    rate_percent = float(input())
    rate = rate_percent / 100  
    
    print("Enter the time:")
    time = float(input())
    
    interest = principal * rate * time
    total_balance = principal + interest    
    
    print(f"The intrest is {interest}, and the balance is {total_balance}")
calulate_simple_interest()
```

```python
#Write a program that calculates the amount of paint needed to cover a wall, given its height and width in meters (assume 1 liter of paint covers 10 square meters).
def calulate_paint():
    print("Enter the width")
    width = int(input())
    print("Enter the height")
    height = int(input())
    area = width * height
    paint_required = area/10
    
    print(f"The amount of paint needed {paint_required}")
calulate_paint()
```

```python
#Create a program that takes a three-digit number and calculates the sum of its digits.
def calulate_sum():
    print("Enter the number")
    num1 = input()
    total = 0
    for i in num1:
        x = int(i)
        total = total+x
    
    
    print(f"The sum is:{total}")
calulate_sum()
```

```python
#Write a program that calculates the hypotenuse of a right triangle based on the lengths of the other two sides entered by the user.
def calulate_hypotenus():
    print("Enter the first number")
    num1 = int(input())
    print("Enter the second number")
    num2 = int(input())
    num1 = num1**2
    num2 = num2**2
    total = num1+num2
    hypotenus = total**0.5
    
    print(f"The hypotenus is:{hypotenus}")
calulate_hypotenus()
```

```python
#Write a program that takes a base salary and calculates net pay after deducting taxes (assume a fixed tax rate of 20%).
def calulate_net_pay():
    TAX_RATE=20
    print("Enter the base salary")
    salary = int(input())
    net_pay = salary* (1- TAX_RATE/100)
    
    print(f"The net pay is:{net_pay}")
calulate_net_pay()
```

```python
#Write a program that takes a price without tax and calculates both the tax amount (at 7.5%) and the final price.
def calulate_total_price():
    TAX_RATE=7.5
    print("Enter the base salary")
    price = int(input())
    tax_amount = price * (TAX_RATE / 100)
    final_price = price + tax_amount
    
    print(f"The tax amount is:{tax_amount} and the final amount is {final_price}")
calulate_total_price()
```
